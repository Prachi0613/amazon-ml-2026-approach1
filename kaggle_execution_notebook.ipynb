## 4. Inspect Real `train_ground_truth.tsv` Header & `matched_entity_ids` Serialization
We need to definitively verify the delimiter used in the real Kaggle ground truth before we trust our candidate-recall metrics.

```python
import pandas as pd

dataset_dir = "/kaggle/input/datasets/lokeshgile/student-resource-amazonml/student_resource/dataset"
gt_path = f"{dataset_dir}/train_ground_truth.tsv"

# Read the first 10 rows to inspect
df_gt = pd.read_csv(gt_path, sep="\t", nrows=10)
print("Ground Truth Columns:", df_gt.columns.tolist())
print("\nFirst 10 Ground Truth Rows:")
print(df_gt.to_string())

# Inspect non-empty matched_entity_ids specifically
sample_matches = df_gt[df_gt["matched_entity_ids"].notna() & (df_gt["matched_entity_ids"] != "")]["matched_entity_ids"].head(10).tolist()
print("\nSample matched_entity_ids values:")
for val in sample_matches:
    print(repr(val))
```
